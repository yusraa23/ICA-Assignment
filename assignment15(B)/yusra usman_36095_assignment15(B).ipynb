```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_tea_snacks.csv")

df.head()
```

```python
df.columns
```

```python
tea_counts = df.groupby("tea_type")["student_id"].count()

print(tea_counts)

print("Most preferred tea:", tea_counts.idxmax())
```

```python
average_cups = df.groupby("gender")["cups_per_day"].mean()

print(average_cups)
```

```python
snack_counts = df.groupby("snack_choice")["student_id"].count()


plt.bar(snack_counts.index, snack_counts.values)

plt.title("Student Snack Preferences")
plt.xlabel("Snack Choices")
plt.ylabel("Number of Students")

plt.show()
```

```python
age_tea = df.groupby(["age_group", "tea_type"])["student_id"].count()

print(age_tea)
```

```python
age_tea.plot(kind="bar")

plt.title("Tea Preference by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")

plt.show()
```

```python
non_tea_drinkers = (df["prefers_tea"] == "No").sum()

print("Students who don't drink tea:", non_tea_drinkers)
```
