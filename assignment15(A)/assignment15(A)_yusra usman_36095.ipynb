```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ride_hailing_dataset.csv")

df.head()
```

```python
df.columns
```

```python
average_fare = df.groupby("service_type")["fare"].mean()

print(average_fare)
```

```python
city_trips = df["city"].value_counts()

print(city_trips)
```

```python
print(city_trips.idxmax())
```

```python
average_distance = df["distance_km"].mean()

print(average_distance)
```

```python
gender_rating = df.groupby("gender")["trip_rating"].mean()

print(gender_rating)
```

```python
average_fare = df.groupby("service_type")["fare"].mean()

plt.bar(average_fare.index, average_fare.values)

plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare")

plt.show()
```
